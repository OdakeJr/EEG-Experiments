# Load MI Dataset

```python
import mne
import numpy as np
from pathlib import Path

def load_bci_iv_2a_gdf(gdf_path):
    """
    Load a BCI Competition IV 2a GDF file.
    """
    raw = mne.io.read_raw_gdf(
        gdf_path,
        preload=True,
        stim_channel="auto",
        verbose=False
    )

    # Keep EEG + EOG separated
    eeg_channels = raw.ch_names[:22]
    eog_channels = raw.ch_names[22:25]

    raw.pick_channels(eeg_channels + eog_channels)

    events, event_id = mne.events_from_annotations(raw)

    return raw, events, event_id
```

```python
def load_bci_iv_2a_dataset(root_dir):
    root_dir = Path(root_dir)
    dataset = {}

    for subj in range(1, 10):
        subj_id = f"A{subj:02d}"

        dataset[subj_id] = {}

        for split, suffix in [("train", "T"), ("test", "E")]:
            gdf_file = root_dir / f"{subj_id}{suffix}.gdf"

            raw, events, event_id = load_bci_iv_2a_gdf(gdf_file)

            dataset[subj_id][split] = {
                "raw": raw,
                "events": events,
                "event_id": event_id
            }

            #print(
            #    f"Loaded {subj_id}{suffix} | "
            #    f"EEG shape: {raw.get_data(picks='eeg').shape}"
            #)

    return dataset
```

```python
data = load_bci_iv_2a_dataset("../../../Datasets/MotorImagery/original/BCICIV_2a_gdf/")
```

```python
raw = data["A01"]["train"]["raw"]
events = data["A01"]["train"]["events"]
event_id = data["A01"]["train"]["event_id"]
```

```python
import numpy as np
import mne
from pathlib import Path
from scipy.io import loadmat

# === Paths ===
root = Path("../../../Datasets/MotorImagery/original/BCICIV_2a_gdf/")  # GDF files
root_mat = Path("../../../Datasets/MotorImagery/original/true_labels/")  # .mat label files

# === Basic info ===
mi_codes = ['769', '770', '771', '772', '783']
#code_map = {769: 0, 770: 1, 771: 2, 772: 3}  # map to 0–3

# === Dictionary to store everything ===
all_data = {}

for subj in range(1, 10):
    subj_id = f"A{subj:02d}"
    all_data[subj_id] = {}

    for sess, suffix in [("session1", "T"), ("session2", "E")]:
        gdf_path = root / f"{subj_id}{suffix}.gdf"
        mat_path = root_mat / f"{subj_id}{suffix}.mat"

        #print(f"Loading {gdf_path.name} ...")

        # --- Load raw EEG ---
        raw = mne.io.read_raw_gdf(gdf_path, preload=True, verbose=False)
        events, event_id = mne.events_from_annotations(raw)

        # --- Select motor imagery cues ---
        mi_event_id = {k: event_id[k] for k in event_id if k in mi_codes}
        if not mi_event_id:
            print(f"  ⚠️ No MI events found in {gdf_path.name}")
            continue

        # --- Epoch 0–4 s after cue ---
        epochs = mne.Epochs(
            raw,
            events,
            event_id=mi_event_id,
            tmin=0.0,
            tmax=4.0,
            baseline=None,
            preload=True,
            verbose=False,
        )

        X = epochs.get_data()  # shape: (trials, channels, samples)

        # --- Load labels from .mat file ---
        labels = loadmat(mat_path)["classlabel"].squeeze()
        y = labels - 1  # convert to 0–3 integers

        all_data[subj_id][sess] = {"X": X, "y": y}

print("\n✅ Finished loading all subjects and sessions.")
```

```python
from collections import Counter

print("Label distribution per subject and session:\n")

for subj_id, sessions in all_data.items():
    print(f"🔹 {subj_id}")
    for sess_name, sess_data in sessions.items():
        y = sess_data["y"]
        counts = Counter(y)
        total = len(y)
        print(f"  {sess_name:9s} | total={total:3d} | " +
              " ".join([f"class {k}: {v:3d}" for k, v in sorted(counts.items())]))
    print()
```
