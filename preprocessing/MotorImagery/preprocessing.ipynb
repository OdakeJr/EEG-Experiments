# Load and structure in samples and labels

```python
import numpy as np
import mne
from pathlib import Path
from scipy.io import loadmat

# === Paths ===
root = Path("../../Datasets/MotorImagery/original/BCICIV_2a_gdf/")  # GDF files
root_mat = Path("../../Datasets/MotorImagery/original/true_labels/")  # .mat label files

# === Basic info ===
mi_codes = ['769', '770', '771', '772', '783']
#code_map = {769: 0, 770: 1, 771: 2, 772: 3}  # map to 0–3

# === Dictionary to store everything ===
all_data = {}

for subj in range(1, 10):
    subj_id = f"A{subj:02d}"
    all_data[subj_id] = {}

    for sess, suffix in [("session1", "T"), ("session2", "E")]:
        gdf_path = root / f"{subj_id}{suffix}.gdf"
        mat_path = root_mat / f"{subj_id}{suffix}.mat"

        #print(f"Loading {gdf_path.name} ...")

        # --- Load raw EEG ---
        raw = mne.io.read_raw_gdf(gdf_path, preload=True, verbose=False)
        ch_names = raw.info["ch_names"]
        eeg_names = ch_names[:22]
        raw.pick_channels(eeg_names)
        events, event_id = mne.events_from_annotations(raw)

        # --- Select motor imagery cues ---
        mi_event_id = {k: event_id[k] for k in event_id if k in mi_codes}
        if not mi_event_id:
            print(f"  ⚠️ No MI events found in {gdf_path.name}")
            continue

        # --- Epoch 0–4 s after cue ---
        epochs = mne.Epochs(
            raw,
            events,
            event_id=mi_event_id,
            tmin=0.5,
            tmax=3.5,
            baseline=None,
            preload=True,
            verbose=False,
        )

        X = epochs.get_data()  # shape: (trials, channels, samples)

        # --- Load labels from .mat file ---
        labels = loadmat(mat_path)["classlabel"].squeeze()
        y = labels - 1  # convert to 0–3 integers

        all_data[subj_id][sess] = {"X": X, "y": y}

print("\n✅ Finished loading all subjects and sessions.")
```

# Filters and Feature extractions

```python
from scipy.signal import butter, sosfiltfilt, resample
import numpy as np
from copy import deepcopy

def band_filter(data, fs=250.0, band=(4, 40), order=5):
    """Band-pass filter along the last axis (time)."""
    nyq = fs / 2.0
    low, high = band[0] / nyq, band[1] / nyq
    sos = butter(order, [low, high], btype="band", output="sos")
    return sosfiltfilt(sos, data, axis=-1)

def downsample_epochs(X, new_fs=128, old_fs=250):
    """Downsample each epoch along the time axis."""
    n_new = int(X.shape[-1] * new_fs / old_fs)
    return resample(X, n_new, axis=-1)

# --- Apply filter and downsample to all subjects/sessions ---
filtered_data = deepcopy(all_data)   # keep structure and labels

for subj_id, sessions in all_data.items():
    for sess_name, sess_data in sessions.items():
        X = sess_data["X"]
        y = sess_data["y"]

        # 1️⃣ Band-pass filter (4–40 Hz)
        X_filt = band_filter(X, fs=250.0, band=(4, 40), order=5)

        # 2️⃣ Downsample (250 → 128 Hz)
        #X_ds = downsample_epochs(X_filt, new_fs=64, old_fs=250)
        X_ds = X_filt

        filtered_data[subj_id][sess_name]["X"] = X_ds
        filtered_data[subj_id][sess_name]["y"] = y

print("✅ All epochs filtered (4–40 Hz) and downsampled (250→128 Hz) in `filtered_data`.")
```

```python
import pandas as pd
import numpy as np
from numpy.fft import fft
from tqdm import tqdm

def compute_time_cov(matrix):
    return np.cov(matrix)

def compute_freq_cov(matrix):
    fft_vals = np.abs(fft(matrix, axis=0))
    return np.cov(fft_vals)

def flatten_covariance(cov, prefix):
    idx = np.triu_indices_from(cov)
    vals = cov[idx]
    names = [f"{prefix}{i}_{j}" for i, j in zip(idx[0], idx[1])]
    return vals, names


# --- Feature extraction loop ---
all_features = []

for subj_id, sessions in tqdm(filtered_data.items(), desc="Subjects"):
    for sess_name, sess_data in sessions.items():
        X = sess_data["X"]
        y = sess_data["y"]

        for trial_idx, trial in enumerate(X):
            # Covariance (time)
            cov_t = compute_time_cov(trial)
            cov_t_vals, cov_t_names = flatten_covariance(cov_t, prefix="time_")

            # Covariance (frequency)
            cov_f = compute_freq_cov(trial)
            cov_f_vals, cov_f_names = flatten_covariance(cov_f, prefix="freq_")

            # Combine both
            feature_vals = np.concatenate([cov_t_vals, cov_f_vals])
            feature_names = cov_t_names + cov_f_names

            # Append row
            all_features.append({
                "subject": subj_id,
                "session": sess_name,
                "label": int(y[trial_idx]),
                **{feature_names[i]: feature_vals[i] for i in range(len(feature_vals))}
            })

print("✅ Feature extraction complete!")
# --- Create one big DataFrame ---
df_features = pd.DataFrame(all_features)

print(df_features.shape)
df_features.head()
```
